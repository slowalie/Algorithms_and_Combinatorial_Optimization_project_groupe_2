For my example, i choose any random file solomon

```python
# Section 3.5 - Formalised example: load the instance C201 (file C201.txt in the same folder)
import pandas as pd
import matplotlib.pyplot as plt

# 1. Read the instance file (Solomon format)
lines = [ln.split() for ln in open("C201.txt") if ln.strip()]
rows = [[float(x) for x in r] for r in lines if len(r) == 7 and r[0].isdigit()]
df = pd.DataFrame(rows, columns=["ID", "X", "Y", "Demand", "Ready", "Due", "Service"]).astype(int)

# 2. Table: depot (ID 0) and the first 9 customers
display(df.head(10))

# 3. Basic graph: depot in red, customers in blue
plt.figure(figsize=(6, 5))
plt.scatter(df.X[1:], df.Y[1:], c="tab:blue", label="customers")
plt.scatter(df.X[0], df.Y[0], c="tab:red", marker="s", s=100, label="depot")
plt.title("Instance C201 (Solomon, 100 customers)")
plt.xlabel("x"); plt.ylabel("y"); plt.legend()
plt.show()
```
