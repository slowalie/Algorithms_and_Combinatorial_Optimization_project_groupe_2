# Formal definition

The **Capacitated Vehicle Routing Problem with Time Windows (CVRPTW)** consists of finding routes that serve every customer exactly once, respect vehicle capacities, customer demands, time windows, travel times, and service durations, and minimize the total travel cost.

We model the problem on a **directed graph**:

$$
G=(V,A),\qquad V=\{0\}\cup N,\qquad
A=\{(i,j)\in V\times V:i\ne j\}.
$$

Node $0$ is the depot, $N$ is the set of customers, and an arc $(i,j)\in A$ represents a possible direct trip from $i$ to $j$. Each used vehicle leaves the depot, visits its assigned customers, and returns to the depot.

**Assumptions:** All vehicles have capacity $Q$; each customer is served entirely by one vehicle (no split deliveries); each vehicle performs at most one route; waiting before service is allowed; travel times, service durations, and time-window bounds are given in **integer time units** (for example, minutes). The last assumption permits a **pure integer linear programming (ILP)** formulation without Big-M. If the source data use fractions of a minute, first choose a sufficiently fine exact integer time unit or acknowledge the effect of discretization.

The depot operates during $[b_0,e_0]$: vehicles depart no earlier than $b_0$ and must return no later than $e_0$. Since waiting is allowed and departure time does not contribute to the objective, departures may be taken at $b_0$ without loss of feasibility.

## Variables

- **Route decision:** $x_{ijk}\in\{0,1\}$ equals $1$ if vehicle $k$ directly travels from node $i$ to node $j$, and $0$ otherwise; $(i,j)\in A$, $k\in K$.
- **Customer assignment:** $y_{ik}\in\{0,1\}$ equals $1$ if vehicle $k$ serves customer $i$, and $0$ otherwise; $i\in N$, $k\in K$.
- **Vehicle usage:** $z_k\in\{0,1\}$ equals $1$ if vehicle $k$ is used, and $0$ otherwise; $k\in K$.
- **Service-start time selection:** $h_{itk}\in\{0,1\}$ equals $1$ if vehicle $k$ starts service at customer $i$ at time $t$, and $0$ otherwise; $i\in N$, $t\in H_i$, $k\in K$.

The feasible service-start slots for customer $i$ are **known before optimization**:

$$
H_i=\{t\in\mathbb Z:b_i\le t\le e_i\}.
$$

The service-start time from the original formulation is **recovered** as

$$
T_{ik}=\sum_{t\in H_i}t\,h_{itk}.
$$

Thus $T_{ik}$ is automatically integer and equals $0$ if $y_{ik}=0$. It is not necessary to introduce $T_{ik}$ as an additional continuous variable. We also avoid MTZ ordering variables: subtours are eliminated by direct **subtour-elimination constraints**.

## Inputs

| Symbol | Meaning | Unit |
|---|---|---|
| $V$ | All nodes: the depot and customers, $V=\{0\}\cup N$ | — |
| $A$ | Directed arcs $(i,j)$ with $i\ne j$ | — |
| $N$ | Set of customers; $n=\lvert N\rvert$ | — |
| $K$ | Set of available vehicles | — |
| $d_i$ | Demand of customer $i\in N$ | Volume |
| $Q$ | Capacity of each vehicle | Volume |
| $c_{ij}$ | Travel cost from node $i$ to node $j$ | Cost |
| $\tau_{ij}$ | Travel time from node $i$ to node $j$ | Integer time units |
| $s_i$ | Service duration at node $i$; $s_0=0$ | Integer time units |
| $[b_i,e_i]$ | Allowed start-of-service time window at customer $i$ | Integer time units |
| $[b_0,e_0]$ | Depot operating time window | Integer time units |
| $H_i$ | Feasible integer service-start times for customer $i$ | Integer time slots |

To minimize **total travel time**, as in the original document, take $c_{ij}=\tau_{ij}$.



## Outputs

A solution provides:

- The ordered route of each used vehicle, for example $R_k=(0,i_1,i_2,\ldots,i_r,0)$.
- The vehicle serving each customer, indicated by $y_{ik}$.
- The start of service at each customer, obtained from $h_{itk}$ or $T_{ik}=\sum_{t\in H_i}t\,h_{itk}$.
- The minimum total travel cost $Z^*$ among all feasible routes.

## Constraints

The model satisfies the **same nine requirements** as the initial formulation:

1. **Exactly one service per customer:** Each customer is assigned to exactly one vehicle.
2. **Flow conservation:** A vehicle that serves a customer both enters and leaves that customer once.
3. **Depot departure and return:** Each used vehicle departs once and returns once; unused vehicles perform no route.
4. **Vehicle capacity:** The total demand carried by each vehicle does not exceed $Q$.
5. **Time windows:** Customer service starts within $[b_i,e_i]$.
6. **Travel and service durations:** The next service cannot start before the previous service and travel are complete.
7. **Return deadline:** Every used vehicle returns no later than $e_0$.
8. **No subtours:** Every served customer belongs to a route connected to the depot.
9. **Variable domains:** All decision variables are binary (hence integer).

## Mathematical formulation

Let $n=\lvert N\rvert$. All inequalities below are **linear** and contain **no Big-M**.

### 1. Objective: minimize total travel time

$$
\boxed{\min Z=\sum_{k\in K}\sum_{(i,j)\in A}\tau_{ij}x_{ijk}.}
$$

(Replace $\tau_{ij}$ by $c_{ij}$ if the objective is monetary cost instead.)

### 2. Visit every customer exactly once

$$
\boxed{\sum_{k\in K}y_{ik}=1\qquad\forall i\in N.}
$$

### 3. Link each customer to its incoming and outgoing arcs (flow conservation)

For every customer $i$ and vehicle $k$:

$$
\boxed{\sum_{j\in V\setminus\{i\}}x_{ijk}=y_{ik}
\qquad\forall i\in N,\ k\in K,}
$$

$$
\boxed{\sum_{j\in V\setminus\{i\}}x_{jik}=y_{ik}
\qquad\forall i\in N,\ k\in K.}
$$

These constraints require exactly one incoming and one outgoing arc for the vehicle serving customer $i$. They also forbid any arc through a customer not assigned to that vehicle.

### 4. Start and end each used route at the depot

$$
\boxed{\sum_{j\in N}x_{0jk}=z_k
\qquad\forall k\in K,}
$$

$$
\boxed{\sum_{i\in N}x_{i0k}=z_k
\qquad\forall k\in K.}
$$

Since $z_k$ is binary, each vehicle is used at most once. If every available vehicle must be used, set $z_k=1$ for all $k\in K$.

### 5. Vehicle capacity

$$
\boxed{\sum_{i\in N}d_i y_{ik}\le Qz_k
\qquad\forall k\in K.}
$$

Each route independently satisfies the vehicle capacity. A necessary (not sufficient) feasibility condition is $\sum_{i\in N}d_i\le\lvert K\rvert Q$.

### 6. Customer time windows: choose one admissible time slot

$$
\boxed{\sum_{t\in H_i}h_{itk}=y_{ik}
\qquad\forall i\in N,\ k\in K.}
$$

Because $H_i=\{b_i,b_i+1,\ldots,e_i\}$, a served customer gets exactly one service-start time within its time window. An unserved customer has no selected time slot for that vehicle.

Equivalently, the recovered integer times satisfy

$$
b_i y_{ik}\le T_{ik}\le e_i y_{ik},
\qquad T_{ik}=\sum_{t\in H_i}t\,h_{itk}.
$$

### 7. Travel time and chronological consistency

For any two distinct customers $i,j$, define an **incompatible pair of start times** by

$$
t' < t+s_i+\tau_{ij},\qquad t\in H_i,\ t'\in H_j.
$$

Such a pair cannot occur when the vehicle travels directly from $i$ to $j$. Exclude every incompatible combination using:

$$
\boxed{x_{ijk}+h_{itk}+h_{jt'k}\le 2}
$$

$$
\forall k\in K,\quad\forall i,j\in N\ (i\ne j),\quad
\forall(t,t')\in H_i\times H_j\text{ such that }t'<t+s_i+\tau_{ij}.
$$

**Explanation:** If the arc $i\to j$ is selected and the starts at $t$ and $t'$ are both selected, the left side would be $3$, which is forbidden whenever there is not enough time for service at $i$ and travel to $j$. For a valid pair, no restriction is added. This implements time precedence using only binary variables and a constant $2$, **not a Big-M relaxation**.

**First customer after the depot.** The vehicle cannot start serving customer $j$ earlier than $b_0+\tau_{0j}$. Exclude infeasible first-customer times:

$$
\boxed{x_{0jk}+h_{jtk}\le1}
$$

$$
\forall j\in N,\ k\in K,\quad
\forall t\in H_j\text{ such that }t<b_0+\tau_{0j}.
$$

Vehicles may arrive early and wait until an admissible start-of-service time.

### 8. Return to the depot before closing

A vehicle cannot make customer $i$ its last stop at time $t$ if it would return after $e_0$. Exclude every such combination:

$$
\boxed{x_{i0k}+h_{itk}\le1}
$$

$$
\forall i\in N,\ k\in K,\quad
\forall t\in H_i\text{ such that }t+s_i+\tau_{i0}>e_0.
$$

Hence, whenever $x_{i0k}=1$, the selected service-start time must satisfy $T_{ik}+s_i+\tau_{i0}\le e_0$.

### 9. Eliminate subtours: classical subtour-elimination constraints (SEC)

For each vehicle, prevent a closed cycle involving only customers:

$$
\boxed{\sum_{i\in S}\sum_{\substack{j\in S\\j\ne i}}x_{ijk}\le |S|-1}
$$

$$
\forall k\in K,\qquad\forall S\subseteq N\text{ with }|S|\ge2.
$$

A cycle visiting all nodes of some customer subset $S$ would have $|S|$ selected arcs internal to $S$, contradicting the bound $|S|-1$. This also rules out cycles with zero total travel or service time. No MTZ variables or MTZ/Big-M-style precedence inequalities are required.

### 10. Variable domains: pure binary ILP

$$
\boxed{x_{ijk}\in\{0,1\}\quad\forall(i,j)\in A,\ k\in K,}
$$

$$
\boxed{y_{ik}\in\{0,1\}\quad\forall i\in N,\ k\in K,}
$$

$$
\boxed{z_k\in\{0,1\}\quad\forall k\in K,}
$$

$$
\boxed{h_{itk}\in\{0,1\}\quad\forall i\in N,\ t\in H_i,\ k\in K.}
$$

All the decision variables are **binary integers**. The start times $T_{ik}$ are integer expressions derived from these variables, not continuous variables.

## ILP statement (compact version)

**Minimize:**

$$
\min\ \sum_{k\in K}\sum_{(i,j)\in A}\tau_{ij}x_{ijk}.
$$

**Subject to:**

$$
\begin{aligned}
\sum_{k\in K}y_{ik}&=1
&&\forall i\in N,\\
\sum_{j\in V\setminus\{i\}}x_{ijk}&=y_{ik}
&&\forall i\in N,\ k\in K,\\
\sum_{j\in V\setminus\{i\}}x_{jik}&=y_{ik}
&&\forall i\in N,\ k\in K,\\
\sum_{j\in N}x_{0jk}&=z_k
&&\forall k\in K,\\
\sum_{i\in N}x_{i0k}&=z_k
&&\forall k\in K,\\
\sum_{i\in N}d_i y_{ik}&\le Qz_k
&&\forall k\in K,\\
\sum_{t\in H_i}h_{itk}&=y_{ik}
&&\forall i\in N,\ k\in K.
\end{aligned}
$$

Together with the following **conditional-incompatibility cuts**, **subtour cuts**, and **binary domains**:

$$
\begin{aligned}
x_{ijk}+h_{itk}+h_{jt'k}&\le2
&&\forall i\ne j\in N,\ k\in K,\ t\in H_i,\ t'\in H_j:\ t'<t+s_i+\tau_{ij},\\
x_{0jk}+h_{jtk}&\le1
&&\forall j\in N,\ k\in K,\ t\in H_j:\ t<b_0+\tau_{0j},\\
x_{i0k}+h_{itk}&\le1
&&\forall i\in N,\ k\in K,\ t\in H_i:\ t+s_i+\tau_{i0}>e_0,\\
\sum_{i\in S}\sum_{\substack{j\in S\\j\ne i}}x_{ijk}&\le|S|-1
&&\forall S\subseteq N:\ |S|\ge2,\ k\in K,\\
x_{ijk},\ y_{ik},\ z_k,\ h_{itk}&\in\{0,1\}
&&\text{on their respective index sets.}
\end{aligned}
$$

**Result:** This is a **pure 0–1 ILP formulation** of the CVRPTW, under the stated discrete-time assumption. It preserves exactly-once service, customer flow, depot departure/return, vehicle capacities, time windows, service and travel durations, the depot closing time, and subtour elimination.

# **Theorical study of the problem**

## 1. Complexity classification

The **Capacited Vehicle Routing Problem with Time Windows(CVRPTW)** is an NP-hard combinatorial problem.

The objective is to Minimize the total routing time cost while satisfying vehicle capacity constraints and customer time-window constraints. Every customer must be visited exactly once, and every vehicle route must start and end at the depot.

To establish its complexity, we consider the associated decision problem.

**CVRPTW-Decision**: Given a CVRPTW istance and cost bound $B$,does there exist a set of feasible vehicle routes whose total cost is at most $B$ ?
The complexity classification is:
- Decision version: **NP-complete**.
- Optimization version: **NP-hard**.

We establish this classification by proving that the decision version belongs to NP ad NP-hard through a polynomial-time reduction from the decision version of the traveling salesman problem (TSP).

## 2. Proof that CVRPTW-Decision belongs to NP

To prove that CVRPTW-Decision belongs to NP,we must demonstrate that a proposed solution can be verified in polynomial time.

### Certificate

A certificate is a set of vehicule routes representing a solution structure for our problem:

$$
S=(R_1,R_2,\ldots,R_m)
\qquad m\leq |K|
$$

Where each route starts and ends at the depot and specifies the sequence of customers visited by its vehicle.

### Verification

Given a certificate $S$ as a solution, must verify the following conditions based on the problem constraints.

1. **Customer coverage:** Every customer is visited exactly once.
2. **Route structure:** Every vehicle route starts and eds at the depot, and the number of routes does not exceed the available fleet size.
3. **Capacity constraints:** For each route $R_k$, the total demad of its customers does not exceed the vehicle capacity $Q$
4. **Time-window constraints:** Each customer is served within its specified time window.$O(1)$
5. **Time cost constraints:** The total cost of all routes is at
must $B$.$O(1)$
6. **Depot return deadline:** every used route returns by time $e_0$.

For time-window verification, let customer $i$ have a time window $[b_i,e_i]$, service duration $s_i$, and let $t_{ij}$ the travel time form customer $i$ to customer $j$.

The earliest feasible service-start time at customer $j$, following customer $i$, is:

$$
T_j=\max\left(b_j,T_i+s_i+\tau_{ij}\right)
$$

When then verify that $T_j\leq b_j$.
If this enequality fails for any customer, the proposed route is infeasible. Otherwise, we continue along the route, also checking ay appicable depot return deadline.

All these checks require only polynomial time in the input size. For example, with direct access to the cost and travel-time matrices, the routes can be checked using a number of operations polynomial in the number of customers and vehicles.

Therefore,

$$
\boxed{\text{CVRPTW-Decision}\in NP}
$$

## 3. Proof that CVRPTW-Decision is NP-Hard

We are going to prove this by reducing the decision problem of the travelling salesman problem to CVRPTW-Decision.

### 3.1. Source problem: TSP-Decision

An instance of TSP-Decision consists of a complete weighted graph $G=(V,E)$
a depot vertex $r\in V$, edhe time costs $t_{ij}$ and costs boud $B$.

The question is whether there exists a hamiltoniann cycle that starts at $r$, visits every vertx exacly once, returs to $r$, ad has total time cost at most $B$.

TSP-Decision is NP-Complete.

### 3.2. Polyomial-time trasormation

Let $m=|V|$. Given an arbitrary TSP-Decision instance, construct a CVRPTW instance as follows:

- **Depot:** use the TSP vertex $r$ as the depot, relabeled $0$.
- **Customers:** each vertex in $V\setminus\{r\}$ becomes a customer.
- **Fleet size:** provide exactly **one** vehicle.
- **Demand:** set $d_i=1$ for every customer.
- **Vehicle capacity:** set $Q=m-1$, sufficient to serve all customers.
- **Travel costs:** preserve every TSP arc cost, $c'_{ij}=c_{ij}$.
- **Travel times:** set $\tau'_{ij}=c_{ij}$ so that total travel time equals TSP tour cost.
- **Service times:** set $s_i=0$ for every customer and for the depot.
- **Time windows:** assign $[0,H]$ to every customer and to the depot, with

$$
H=mC_{\max}+1,\qquad C_{\max}=\max_{(i,j)\in E}c_{ij}.
$$

A Hamiltonian cycle has exactly $m$ arcs, each costing at most $C_{\max}$; therefore its total travel time is at most $mC_{\max}<H$. These wide windows do not remove any TSP tour.

- **Objective bound:** keep the same bound $B$.

The constructed instance contains a polynomial number of nodes and arcs. Because the integer $H$ has polynomially many bits relative to the input encoding, this transformation runs in polynomial time.

### 3.3. Proof of reduction

We prove equivalence in both directions.

**Direction 1: TSP-Decision has a solution $\Rightarrow$ CVRPTW-Decision has a solution.**

Suppose that the TSP instance has a Hamiltonian cycle of total cost at most $B$. Use this cycle as the route of the single vehicle. Every customer is visited exactly once; capacity $Q=m-1$ allows all unit demands; service durations are zero; and the time horizon is large enough to accommodate the complete route. Since travel times equal the original TSP costs, the total CVRPTW travel time is also at most $B$. Therefore, the constructed CVRPTW instance is a **yes-instance**.

**Direction 2: CVRPTW-Decision has a solution $\Rightarrow$ TSP-Decision has a solution.**

Suppose that the constructed CVRPTW instance admits a feasible solution of total travel time at most $B$. There is only one vehicle, and the routing rules require it to start from the depot, serve every customer exactly once and return to the depot. Consequently, its route is a Hamiltonian cycle through all vertices of the original TSP graph. Because the travel times were copied from TSP arc costs, the original cycle has cost at most $B$. Therefore, the TSP instance is also a **yes-instance**.

We have established:

$$
\boxed{\text{TSP-Decision is a yes-instance}\iff\text{CVRPTW-Decision is a yes-instance}.}
$$

Hence:

$$
\text{TSP-Decision}\leq_p\text{CVRPTW-Decision}.
$$

Since TSP-Decision is NP-complete, CVRPTW-Decision is NP-hard. Combined with membership in NP, this yields:

$$
\boxed{\text{CVRPTW-Decision is NP-complete}.}
$$

Consequently, the **optimization version of the general CVRPTW is NP-hard**.

## 4. Implications for finding a solution for the problem

The **NP-hardness** of the CVRPTW has several practical consequences:

- **No polynomial-time exact algorithm is expected in the general case unless $P=NP$.** A general-purpose algorithm cannot be assumed to return a proven optimum efficiently for every instance.
- **Exact methods may become computationally expensive.** Mixed-integer optimization can certify optimality on some instances, but the worst-case effort can increase rapidly as problem size grows.
- **Heuristics and metaheuristics are useful for larger instances.** They aim to find feasible, high-quality solutions within an acceptable computation time, without usually proving global optimality.
- **The method should depend on instance size and structure.** Small instances are useful for exact validation, while larger ones motivate methods such as ALNS.
- **Feasibility remains mandatory.** A short route that violates a capacity limit or a time window is not a valid CVRPTW solution.

These implications justify studying both metaheuristic and a scalable heuristic. A computational comparison is still necessary to demonstrate the practical performance of the proposed method.


# Solution Method

##1. Choice of Method and Justification

We propose an Adaptive Large Neighborhood Search (ALNS) approach
for the Capacitated Vehicle Routing Problem with Time Windows
(CVRPTW).

The method combines three components:

1. A feasible insertion heuristic to construct an initial solution.
2. Destroy and repair operators to explore alternative customer
   assignments and visiting sequences.
3. A Simulated Annealing acceptance rule to allow occasional
   non-improving moves during the search.

The objective is to minimize total travel time while serving every
customer exactly once and respecting vehicle capacities, customer
time windows, the available fleet size and the depot return deadline.

## 2. Definition of a Solution

### 2.1 Route representation

Let $N=\{1,\ldots,n\}$ be the set of customers, let $0$ denote
the depot, and let $K$ be the set of available vehicles.

A complete solution is represented by a collection of non-empty
vehicle routes:

$$
S=(R_1,\ldots,R_m),
\qquad 1 \leq m \leq |K|.
$$

Each used vehicle performs one route:

$$
R_k=(v_{k,0},v_{k,1},\ldots,v_{k,r_k},v_{k,r_k+1}),
$$

where:

- $v_{k,0}=v_{k,r_k+1}=0$: the route starts and ends at the depot.
- $v_{k,p}\in N$ for $p=1,\ldots,r_k$: internal nodes are customers.
- $r_k$ is the number of customers served by route $k$.
- $m$ is the number of vehicles used.

Every customer must appear exactly once across all routes:

$$
\sum_{k=1}^{m}\sum_{p=1}^{r_k}
\mathbf{1}\{v_{k,p}=i\}=1,
\qquad \forall i\in N.
$$

Here, $\mathbf{1}\{\cdot\}$ equals 1 when the condition is true
and 0 otherwise.

For example, a solution serving five customers with two vehicles is:

$$
R_1=(0,2,5,0),
\qquad
R_2=(0,1,3,4,0).
$$

The route sequences determine both the assignment of customers
to vehicles and their visiting order. Service-start times are
computed from these sequences using the scheduling procedure
described in the next subsection.



### 2.3 Partial solutions

During the destroy step, some customers are temporarily
removed from the routes. Let $U$ be the set of customers
awaiting reinsertion.

The repair step attempts to reinsert every removed customer
while respecting all constraints.

A candidate solution is considered for acceptance only when
$U=\varnothing$ and all routes are feasible. Otherwise,
the candidate is rejected and the current solution is retained.


## 3. Route Feasibility

The CVRPTW constraints are defined in the Theoretical Study
section. Here, we explain how we check them when inserting
customers into routes.

### 3.1 Feasibility Checks During Insertion

Before adding a customer to a route, we check how the
insertion affects the vehicle's load and schedule:

1. **Vehicle capacity:** We calculate the updated route load
   and check that it stays within the vehicle's capacity.

2. **Service-start times:** We update the schedule along
   the modified route, including travel, service and waiting
   times.

3. **Time windows:** We check the service-start time of every
   customer on the route, since an insertion may also delay
   later visits.

4. **Depot return:** We check that the vehicle can finish
   its deliveries and return before the depot closes.

5. **Fleet availability:** If a new route is needed, we
   check that another vehicle is available.

We keep an insertion only if all checks pass. Otherwise,
we try another position. If no feasible option is available,
the construction or repair procedure handles the failure
as described in the following sections.

We use the same checks to build the initial solution
and to reinsert customers during ALNS repair.



4. Building Initial Solutions

A feasible cheapest insertion approach is used to construct an initial solution.
This heuristic is adapted for the CVRPTW.

At the beginning, no customers are assigned and no routes are opened.

4.1 Insertion costs

If customer $u$ is inserted between two consecutive nodes $i$ and $j$, the travel time increases by: $$ \Delta(u;i,j)=t_{iu}+t_{uj}-t_{ij}. $$

The direct trip from $i$ to $j$ is replaced by the trips from $i$ to $u$ and from $u$ to $j$.

An insertion is possible only when the updated route respects capacity, all customer time windows and the depot return time.


4.2 Construction procedure

1. Find all possible insertions for all unassigned customers for all routes.

2. If there are possible insertions, choose the customer and position that will cause the least increase in travel time.
   Tiebreaker: earliest customer deadline.

3. If no possible insertion is found, create a new route, if the fleet limit is not reached.
   Select the unassigned customer with the earliest deadline for which the single-customer route $(0,u,0)$ is feasible.

4. Update the load and schedule of the modified route, remove the inserted customer from the unassigned set and repeat;

Construction is successful if all customers are assigned.
We call the feasible solution obtained $S_0$.

### 4.3 Breakdown of construction

If customers remain unassigned and no feasible insertion or new route can be performed, the construction attempt fails.

We then do a limited number of restart attempts with randomized choices from feasible insertions and seed customers.
The restart limit and random seed are logged for reproducibility.

If all attempts fail, we report that no feasible initial solution was found in the construction budget.
This does not mean that the instance is infeasible.

### 4.4 Relation to ALNS

The current solution becomes the best solution found so far:

$$
S_{\mathrm{current}}=S_{\mathrm{best}}=S_0.
$$

Then ALNS tries to improve this solution via successive destroy and repair operations.


## 5. Neighborhood Structure and Operators

In ALNS, we explore different solutions by removing some customers from their current routes and inserting them again in new positions. This allows us to change the order of visits or assign customers to different vehicles.

The process can be expressed as:

$$
S'=H(D(S)),
$$

where $D$ represents the destroy operator and $H$ represents the repair operator.

### 5.1 Neighborhood Structure

The neighborhood consists of all feasible solutions that can be obtained by removing and reinserting customers. By doing this, we can explore different route configurations and potentially find a better solution.

At each iteration, we remove $q$ different customers, calculated as:

$$
q=\min(n,\max(1,\lceil\alpha n\rceil)),
\qquad 0<\alpha\leq1,
$$

Here, $\alpha$ determines the proportion of customers removed. For example, if we have 100 customers and choose $\alpha=0.10$, we remove 10 customers before trying to insert them again.

### 5.2 Destroy Operators

We use two methods to remove customers from the current solution:

* **Random removal:** We randomly select $q$ different customers and remove them from their routes. Each customer has an equal chance of being selected.
* **Worst-cost removal:** We remove customers whose removal produces the greatest reduction in travel time.

For worst-cost removal, the saving is calculated as:

$$
g(u)=t_{iu}+t_{uj}-t_{ij},
$$


where $i$ and $j$ are the nodes immediately before and after customer $u$ in the current route. These nodes can be either customers or the depot.

After removing a customer, we recalculate the savings because the route has changed. If a route becomes empty, we delete it.

### 5.3 Repair Operators

After the removal step, we need to insert all removed customers back into the solution. We consider two insertion methods:

* **Greedy insertion:** We look for the customer and insertion position that produce the smallest additional travel cost. We repeat this until all customers have been inserted.
* **Regret-2 insertion:** Instead of focusing only on the cheapest insertion, we also consider how much more expensive the second-best option would be.

The regret value is calculated as:

$$
r(u)=c_2(u)-c_1(u),
$$

where $c_1(u)$ is the cheapest feasible insertion cost and $c_2(u)$ is the second-cheapest.

A customer with a high regret value is inserted first because delaying that insertion could lead to a much higher cost later. Once selected, the customer is placed in its cheapest feasible position.

If a customer has only one feasible insertion position, we give it priority because there is no alternative position available. If several customers have only one feasible option, we select the one with the lowest insertion cost first.

After each insertion, we update the insertion costs and feasible positions for the remaining customers, since the routes may have changed.

### 5.4 Feasibility and Repair Failure

When inserting customers, we must make sure that all constraints are respected. These include vehicle capacity, customer time windows, and the deadline for returning to the depot, as explained in Section 3.

If necessary, we can create a new route $(0,u,0)$ for a customer, provided that the route is feasible and we have not reached the maximum number of vehicles.

The cost of opening this route is:

$$
t_{0u}+t_{u0}.
$$

If we cannot find a feasible position for even one removed customer, we reject the candidate solution.

This means that only solutions containing all customers and satisfying every constraint can move on to the acceptance step.

6. Adaptive Operator Selection

In ALNS, we do not always use the same destroy and repair operators. Instead, we select one operator from each group at every iteration. The selection depends on how well each operator has performed in previous iterations.

Each operator $h$ starts with a weight $w_h=1$. Operators with higher weights have a greater chance of being selected.

The selection probability is calculated as:

$$
p_h=\frac{w_h}{\sum_{g\in O}w_g},
$$

where $O$ represents either the destroy or repair operator group.

After each iteration, we evaluate the new solution and give a score to both selected operators based on the result.

| Outcome | Score |
|---------|-------|
| New best solution | 5 |
| Improves the current solution without beating the best | 3 |
| Accepted without improving the current solution | 1 |
| Rejected or infeasible candidate | 0 |

We then update the weights of the selected operators using:

$$
w_h \leftarrow \max\left(\varepsilon, (1-\rho)w_h+\rho\sigma\right)
$$

where $\sigma$ is the score received, $\rho$ controls how quickly the weights change, and $\varepsilon$ ensures that every operator still has a chance of being selected.

Operators that are not selected keep their previous weights.

The main idea is to give more importance to operators that produce good solutions, while still allowing other operators to be explored. This helps the algorithm adapt its search strategy over time.

The scores in the table are the values we propose for our approach. We will test these values and the adaptation parameter during our experiments to see how they affect the results.




## 7. Simulated Annealing Acceptance Rule

When deciding whether to accept a new candidate solution ($S'$) over the current solution ($S$), we only consider solutions that satisfy all the constraints.

### 7.1 Cost Difference

We first calculate the difference between the costs of the two solutions:

$$
\Delta Z = Z(S') - Z(S)
$$

where $Z(S')$ is the cost of the new solution and $Z(S)$ is the cost of the current one.

If $\Delta Z \leq 0$, the new solution is better or has the same cost, so we always accept it.

### 7.2 Acceptance Probability

If the new solution is worse ($\Delta Z > 0$), it can still be accepted with a certain probability.

This probability depends on how much worse the solution is and the current temperature ($T$):

$$
P_{\mathrm{accept}} = \exp\left(-\frac{\Delta Z}{T}\right)
$$

To decide whether to accept a worse solution, we generate a random number $u$ between 0 and 1.

If $u < P_{\mathrm{accept}}$, we accept the new solution even though its cost is higher.

### 7.3 Temperature Cooling

The temperature gradually decreases after each iteration using:

$$
T \leftarrow \max(T_{\min}, \beta T)
$$

where:

- $T_{\min}$ is the minimum temperature.
- $\beta$ is the cooling factor, with $0 < \beta < 1$.

When the temperature is high, the algorithm is more open to accepting worse solutions. This allows us to explore different possibilities and avoid getting stuck in a local optimum.

As the temperature decreases, the algorithm becomes more selective and is less likely to accept worse solutions.

### 7.4 Best Solution and Parameters

The best feasible solution found during the search is stored separately. It is updated whenever a better solution is discovered.

At the end, the algorithm returns the best solution found, rather than simply the last accepted one.

Three parameters need to be evaluated during experimentation:

- **Initial temperature ($T_0$):** The temperature at the beginning of the search.
- **Minimum temperature ($T_{\min}$):** The lowest temperature allowed.
- **Cooling factor ($\beta$):** Controls how quickly the temperature decreases.

These parameters will be tested during our experiments to find suitable values for the algorithm.

## 8. Algorithm Overview and Termination

### 8.1 General procedure

1. Build an initial feasible solution using the insertion heuristic.
   If no feasible starting solution is found within the allowed
   restart attempts, stop and report construction failure.

2. Set the starting solution as both the current and best solution.
   Initialize all operator weights to 1 and set the initial temperature.

3. Select one destroy operator and one repair operator using
   their adaptive selection probabilities.

4. Apply the operators to a copy of the current solution
   to create a new potential solution.

5. Check that the new solution serves every customer exactly once
   and satisfies all CVRPTW constraints. If it does, apply the
   Simulated Annealing rule to decide whether to accept it
   as the current solution.

6. Update the best solution if the new solution has a lower
   cost than the best solution found so far.

7. Update the selected operators' weights based on the outcome.
   Rejected or infeasible solutions receive a score of zero.
   Decrease the temperature.

8. Repeat Steps 3–7 until a stopping criterion is reached.

### 8.2 Stopping criteria

The algorithm stops when either the maximum number of iterations
$I_{\max}$ or the search time limit $t_{\max}$ is reached.

These limits and other parameters are set before each experiment.
Random seeds are recorded to support reproducibility.

### 8.3 Returned solution

The algorithm returns the best feasible solution found, including:

1. Routes for each vehicle.
2. Service-start times at each customer.
3. Total travel time.
4. Load carried by each vehicle.
5. Number of vehicles used.

The final solution satisfies all CVRPTW constraints.
As a metaheuristic, ALNS does not guarantee global optimality.
