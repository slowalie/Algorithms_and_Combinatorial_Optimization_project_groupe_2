# **Formal definition**

**ref:https://www.youtube.com/watch?v=5s-kqPQInpM**
https://www.youtube.com/watch?v=BKcS8aQfT2s
https://how-to.aimms.com/Articles/332/332-Miller-Tucker-Zemlin-formulation.html

The VRPTW Problem consists of finding a set of vehicules routes that serves every customers exaclty once, respects vehicules capacities, customer demand, time widows and sevices times and minimizes the total cost.

We model the problem on an undirected graph

$$G=(V,E), \qquad V=\{0\}\cup N,\qquad E={(i,j)\in V\times V|i\neq j}$$

Where node $0$ is the depot and $N$ is the set of customers. An arc $(i,j)\in E$ represents a posssible trip directly from node $i$ to node $j$. (no split deliberies); and each vehicule starts form the depot visits its assigned customers, and returns to the depot.

**Assumptionns used in this formulation:** vehicules have the same capacity $Q$; each customer's demand is served by a single vehicule (no split deliveries); and each vehicule operates at most oune route. If every available vehicule must be used, the depot constraints can be changed as noted below.

### Variables

- **Route decision variable**:
$$
x_{ijk} = \begin{cases}
1 & \text{if vehicle } k \text{ goes direcly from customer }i \text{ to }j \\
0 & \text{otherwise}
\end{cases}
$$

  Here, $i, j\in V$, $(i,j)\in E$ and $k\in K$

- **Customer assignment variable**: $y_ik=1$ if customer $i$ is served by vehicle k, and 0 otherwise $\forall i\in N, \forall k\in K$.
- **Vhicle usage variable**: $z_k=1$ if vehicle k is used, and 0 otherwise.
- **Time variable**: $T_{ik}$ is the service-start-time at customer i for vehicle k. We set $T_ik=0$ when $y_ik=0$. Arrival may be earlier than $T_ik$ because waiitng is allowed.
- **Subtour-ordering variable**: $u_{ik}\geq 0$ is an auxiliary ordering variable used in the Miller–Tucker–Zemlin (MTZ) constraints to eliminate subtours.


### Inputs

| Symbol | Meaning | Unit |
|---|---|---|
| $V$ | Set of nodes: the depot and all customers, $V = \{0\}\cup N$ |
| $E$| Set of feasible arcs $(i,j)$, with $i\neq j$|
| $N$ | Set of customers; $n=|N|$ is the number of customers|
| $K$ | Set of available vehicules; $|K|$ is the fleet size |
|$d_i$ | Demand of customer $i\in N$|Volume(l, $m^3$)|
| $Q$ | Maximum carrying capacity of each vehicle |Volume(l, $m^3$)|
| $c_{ij}$ | Distance or monetary cost of traveling from $i$ to $j$ |min|
| $\tau_{ij}$| Travel time from node $i$ to node $j$|min|
| $s_i$| service duration at node $i$; take $s_0=0$ for the depot |min|
| $[b_i,e_i]$| Time window for node $i$: service must start no earlier than $b_i$ and no later than $e_i$|min|
| $M$ | A sufficiently large positive constat used in conditional time and return constraints|

The wheight of each vertex is consider as the travel time $c_{ij}=\tau_{ij}$

For the depot, $[b_0,e_0]$ is the operating time window: vehicles may depart no earlier than $b_0$ and must return by $e_0$



### Outputs

A solution provides:

- The route assigned to each vehicle, for example $R_k=(0,i_1,i-2,\ldots,i_r,0)$.
- The assignment of each customer to a vehicle, represented by $y_{ik}$.
- The service-start time $T_{ik}$ for each assigned customer, giving a feasible visit schedule
- The minimum total travel time cost $Z^*$ among all routes satisfying the constraints.

### Constraints

The model must satisfy the following requirements:

1. **Every customer is served exactly once.** Each customer is assigned to one vehicle and is entered and left once by that vehicle.
2. **flow conservation.** A vehicle that is used leaves the depot once and returns once. An unused vehicle has no depot departure or return.
3. **Depot departure and return.** A vehicle that is used leaves the depot once and returns once. An unused vehicle has no depot departure or return.
4. **Vehicle capacity.** the total demand carried by any vehicle canot exceed $Q$.
5. **Time windows.** Service at customer $i$ must start within $[b_i,e_i]$ when a vehicle is assigned to that customer.
6. **Travel and service durations.** If vehicle $k$ travels from $i$ to $j$, service at $j$ cannot start until service at $i$ is completed and the travel time to $j$ has elapsed.
7. **Return deadline.** Each used route must return to the depot no later than $e_0$.
8. **No subtours.** A vehicle can not form a disconnected cycle among customers that does not belong to its depot route.
9. **Variable domains** Route and assigment decisions are binary; time and ordering variables are non-negative.



### Mathematical Formulation
Let $n=|N|$. The complete model is defined by the objective and the constraints below.

#### 1. Objective: minimize total travel cost

$$\boxed{\min Z=\sum_{k\in K}\sum_{i\in V}\sum_{j\in V, j\neq i}\tau_{ij}x_{ijk}} $$

#### 2. Visit every customer exactly once

$$
\boxed{
\sum_{k\in K}y_{ik}=1
\qquad \forall i\in N
}
$$

#### 3. Link customer
for each customer $i$ and vehicle $k$, exactly one outgoing arc and one incoming arc are selected if and only if vehicle $k$ serves that customer:

$$
\sum_{i\in N}\sum_{j\in N}x_{ijk}=\sum_{i\in N}\sum_{j\in N}x_{jik}=y_{ik}
\qquad \forall k\in K
$$


#### 4. Start and end at the depot

Each vehicle can perform at most one route. Its departure and return counts must be equal:

$$
\sum_{j\in N}x_{0jk}=\sum_{i\in N}x_{i0k} = z_k
\qquad \forall k\in K
$$

As $z_k\in \{0,1\}$, both sums are zero when vehicle k is unsued and one when it is used. Not every available vehicle must be used.

#### 5. Vehicle-capacity constraint

The total demand assigned to vehicle k must not exceed its capacity:

$$
\sum_{i\in N}d_iy_{ik}\leq Q
\qquad \forall k\in K
$$

This is **per-vehicle** constraint. The condition $\sum_{i\in N}d_i \leq Q$ would incorrecly require all customers demand to fit in one vehicle. A useful necessary (but not sufficient) check is $\sum_{i\in N}d_i \leq |K|Q$.


#### 6. Customer time windows

When the vehicle $k$ serves customer $i$ he take a certains time $T_{ijk}$ from the rpevious node, the service-start time must lie within the customer's time window:

$$
b_iy_{ik}\leq T_{ik}\leq e_iy_{ik}
\qquad \forall i\in N, \forall k\in K
$$

Because $b_i,e_i\geq 0$, these inequalities enforce $T_{ik}=0$ when $y_{ik}=0$, and $b_i\leq T_{ik}\leq e_i $when $y_{ik}=1. The time window concerns the start of service, not its completion.$

#### 7. Travel time and time consistency

If vehicle $k$ travels from customer $i$ to costomer $j$, the service at $j$ cannot begin before service at $i$ has finished and the travel time has elapsed:

$$
T_{ik}+s_i+\tau_{ij}-M(1-x_{ijk})\leq T_{jk}
\qquad \forall i,j\in N,i\neq j , \forall k\in K
$$

We must also account for travel from the depot to the first customer:

$$
b_0+\tau_{0j}-M(1-x_{0jk})\leq T_{jk}
\qquad \forall j\in N,\forall k\in K
$$

We assume that vehicles can depart at b_0 and wait before serving theri first customer. When $x_{ijk}=0$, the corresponding large-$M$inequalitu is inactive. When $x_{ijk}=1$, it enforces the chronological order of services.

#### 8. Return to the depot before closing

If vehicle k travels from customer $i$ back to the depot, it must finish service and complete the return trip by time $e_0$:

$$
T_{ik}+s_i+\tau_{i0}\leq e_0+M(1-x_{i0k})
\qquad \forall i\in N,\ k\in K
$$


#### 9. Subtour elimination (MTZ constraints)

We use order-based MTS variables, for each customer and vehicle:

$$
\boxed{
  y_{ik}\leq u_{ik}\leq ny_{ik}
  \qquad \forall i\in N,k\in K
}
$$

For two different customers:

$$
\boxed{
  u_{jk}\geq u_{ik}+1-(n+1)(1-x_{ijk})
}
$$

$$
\forall i,j\in N,\ i\neq j,\ k\in K.
$$

If $x_{ijk}=1$, then $u_{jk}\geq u_{ik}+1$: the route position increases strictly along each selected customer-to-customer arc. A closed cycle among customers would require these positions to increase and return to their initial value, which is impossible. When $x_{ijk}=0$, the inequality does not restrict feasible positions.

Time-propagation constraints already eliminate customer cycles when each cycle has strictly positive total travel and service time. MTZ constraints nevertheless provide explicit elimination, including degenerate zero-time cases.
#### 10. Variable domains

$$
x_{ijk}\in\{0,1\}\quad\forall(i,j)\in A,\ k\in K;
\qquad y_{ik}\in\{0,1\}\quad\forall i\in N,\ k\in K;
$$

$$
z_k\in\{0,1\}\quad\forall k\in K;
\qquad T_{ik}\geq0,\ u_{ik}\geq0\quad\forall i\in N,\ k\in K.
$$

The auxiliary ordering variable $u_{ik}$ may be continuous: the MTZ inequalities still prohibit subtours.


### ILP Statement
The VRPTW can be summarized as the following mixed-integer linear optimization model:

**Objective function:**

$$
\min Z=\sum_{k\in K}\sum_{(i,j)\in A}\tau_{ij}x_{ijk}.
$$

**Subject to:**

$$
\begin{aligned}
\sum_{k\in K}y_{ik}&=1 &&\forall i\in N,\\
\sum_{i\in V}\sum_{j\in V}x_{ijk}&=\sum_{i\in V}\sum_{j\in V}x_{jik}=y_{ik} &&\forall i\in N,\ k\in K,\\
\sum_{j\in N}x_{0jk}&=z_k &&\forall k\in K,\\
\sum_{i\in N}x_{i0k}&=z_k &&\forall k\in K,\\
\sum_{i\in N}d_i y_{ik}&\leq Qz_k &&\forall k\in K,\\
b_i y_{ik}\leq T_{ik}&\leq e_i y_{ik} &&\forall i\in N,\ k\in K.
\end{aligned}
$$

$$
\begin{aligned}
T_{jk}&\geq T_{ik}+s_i+\tau_{ij}-M(1-x_{ijk})
&&\forall i\neq j\in N,\ k\in K,\\
T_{jk}&\geq b_0+\tau_{0j}-M(1-x_{0jk})
&&\forall j\in N,\ k\in K,\\
T_{ik}+s_i+\tau_{i0}&\leq e_0+M(1-x_{i0k})
&&\forall i\in N,\ k\in K,\\
y_{ik}\leq u_{ik}&\leq n y_{ik}
&&\forall i\in N,\ k\in K,\\
u_{jk}&\geq u_{ik}+1-(n+1)(1-x_{ijk})
&&\forall i\neq j\in N,\ k\in K.
\end{aligned}
$$
Together with the binary and non-negativity domains in Section 10, these equations describe the same model without introducing a second, contradictory formulation. The model is a **MILP**, not a pure ILP, because $T_{ik}$ and $u_{ik}$ may be continuous.

# Data Representation


## Graph Representation

The Capacitated Vehicle Routing Problem with Time Windows (CVRPTW) is represented by a weighted graph $G=(V,E)$.

The set of vertices is defined as:
 $V = \{0\}\cup N$
Where:

$V$ represents the set of all locations.

0 represents the depot, where vehicles start and return after completing their routes.

$N = {1, 2, 3, ..., n}$ represents the set of customers.

The set of edges $E$ represents the possible movements between locations. We define it as all pairs $(i, j)$ such that $i$ and $j$ belong to $V$ and $i$ is different from $j$.

Each edge $(i, j)$ represents a possible movement from location $i$ to location $j$.

Each edge is associated with two values:

$c_{ij}$: the distance between locations $i$ and $j$.

$\tau_{ij}$: the travel time from location $i$ to location $j$.




## Input parameters

The model requires the following input data.

- $n$, The number of custormers. "Integer"

- $K$, set of available vehicules

- $d_i$, Demand of customers. "Vector containing n+1 values

- $Q$. Maximum capacity of each vehicule

- $(x_i, y_i)$, Coordinate of location $i$. "Matrix with n +1 rows and 2 columns"

- $(b_i, e_i)$, Time window of location

- $s_i$, Service duration at location . "Vector containing n+1 values"

- $c_ij$, Distance between locations $i$ and $j$. "Matrix with n + 1 rows and n + 1 columns"

- $\tau_ij$, travel time between locations $i$ and $j$. "Matrix with n + 1 rows and n + 1 columns"





**Customer demand**
The parameter $d_i$ represents the quantity that must be delivered to customer i. The demand data contains one value for the depot and one value for each customer. We assume that $d_0$ = 0 because the depot does not represent a customer requiring a delivery.



**Vehicule capacity**

The parameter $Q$ represents the maximum quantity that each vehicle can carry. Since all vehicles are assumed to be identical, they have the same capacity.

The total demand assigned to a vehicle must not exceed $Q$.



**Coordinates**
Each location $i$ is represented by its coordinates $(x_i, y_i)$. All coordinates are stored in a matrix with $n + 1$ rows and 2 columns, including the depot and all customers.


**Time windows**

Each location has a time window $(b_i, e_i)$, where:

- $b_i$ is the earliest permitted service start time.

- $e_i$ is the latest permitted service start time.



**Service duration**

The parameter $s_i$ represents the time required to complete the delivery at location i. We assume that $s_0 = 0$, meaning that loading and unloading operations at the depot are not included in the model.



## Decision variables

The following decision variables describe the routes and schedules of the vehicles.

**Route decision variable**
$x_{ijk}\in \{0,1\}$

The variable $x_{ijk}$ equals:

- 1 if vehicle $k$ travels directly from location $i$ to location $j$.

- 0 otherwise.

**Service start time**

$T_{ik}$ ≥ 0

The variable $T_ik$ represents the time at which vehicle $k$ starts serving customer i.

This variable is relevant when customer $i$ is served by vehicle $k$.

## Expected outputs
After solving the CVRPTW instance, the model is expected to provide the following outputs:

- **Vehicle routes**: the sequence of locations visited by each vehicle, starting and ending at the depot.

- **Customer assignments**: the vehicle responsible for serving each customer.

- **Service schedules**: the service start times T_ik, including waiting when a vehicle arrives before a time window opens.

- **Route statistics**: the total distance travelled, the load carried by each vehicle and the number of vehicles used.

- **Feasibility status**: confirmation that every customer is served exactly once, vehicle capacities are respected, time windows are satisfied and vehicles return to the depot before the required deadline.

- **Objective value**:
is to minimize distance.

# **Theorical study of the problem**

## 1. Complexity classification

The **Capacited Vehicle Routing Problem with Time Windows(CVRPTW)** is an NP-hard combinatorial problem.

The objective is to Minimize the total routing time cost while satisfying vehicle capacity constraints and customer time-window constraints. Every customer must be visited exactly once, and every vehicle route must start and end at the depot.

To establish its complexity, we consider the associated decision problem.

**CVRPTW-Decision**: Given a CVRPTW istance and cost bound $B$,does there exist a set of feasible vehicle routes whose total cost is at most $B$ ?
The complexity classification is:
- Decision version: **NP-complete**.
- Optimization version: **NP-hard**.

We establish this classification by proving that the decision version belongs to NP ad NP-hard through a polynomial-time reduction from the decision version of the traveling salesman problem (TSP).

## 2. Proof that CVRPTW-Decision belongs to NP

To prove that CVRPTW-Decision belongs to NP,we must demonstrate that a proposed solution can be verified in polynomial time.

### Certificate

A certificate is a set of vehicule routes representing a solution structure for our problem:

$$
S=(R_1,R_2,\ldots,R_m)
\qquad m\leq |K|
$$

Where each route starts and ends at the depot and specifies the sequence of customers visited by its vehicle.

### Verification

Given a certificate $S$ as a solution, must verify the following conditions based on the problem constraints.

1. **Customer coverage:** Every customer is visited exactly once.
2. **Route structure:** Every vehicle route starts and eds at the depot, and the number of routes does not exceed the available fleet size.
3. **Capacity constraints:** For each route $R_k$, the total demad of its customers does not exceed the vehicle capacity $Q$
4. **Time-window constraints:** Each customer is served within its specified time window.$O(1)$
5. **Time cost constraints:** The total cost of all routes is at
must $B$.$O(1)$
6. **Depot return deadline:** every used route returns by time $e_0$.

For time-window verification, let customer $i$ have a time window $[b_i,e_i]$, service duration $s_i$, and let $t_{ij}$ the travel time form customer $i$ to customer $j$.

The earliest feasible service-start time at customer $j$, following customer $i$, is:

$$
T_j=\max\left(b_j,T_i+s_i+\tau_{ij}\right)
$$

When then verify that $T_j\leq b_j$.
If this enequality fails for any customer, the proposed route is infeasible. Otherwise, we continue along the route, also checking ay appicable depot return deadline.

All these checks require only polynomial time in the input size. For example, with direct access to the cost and travel-time matrices, the routes can be checked using a number of operations polynomial in the number of customers and vehicles.

Therefore,

$$
\boxed{\text{CVRPTW-Decision}\in NP}
$$

## 3. Proof that CVRPTW-Decision is NP-Hard

We are going to prove this by reducing the decision problem of the travelling salesman problem to CVRPTW-Decision.

### 3.1. Source problem: TSP-Decision

An instance of TSP-Decision consists of a complete weighted graph $G=(V,E)$
a depot vertex $r\in V$, edhe time costs $t_{ij}$ and costs boud $B$.

The question is whether there exists a hamiltoniann cycle that starts at $r$, visits every vertx exacly once, returs to $r$, ad has total time cost at most $B$.

TSP-Decision is NP-Complete.

### 3.2. Polyomial-time trasormation

Let $m=|V|$. Given an arbitrary TSP-Decision instance, construct a CVRPTW instance as follows:

- **Depot:** use the TSP vertex $r$ as the depot, relabeled $0$.
- **Customers:** each vertex in $V\setminus\{r\}$ becomes a customer.
- **Fleet size:** provide exactly **one** vehicle.
- **Demand:** set $d_i=1$ for every customer.
- **Vehicle capacity:** set $Q=m-1$, sufficient to serve all customers.
- **Travel costs:** preserve every TSP arc cost, $c'_{ij}=c_{ij}$.
- **Travel times:** set $\tau'_{ij}=c_{ij}$ so that total travel time equals TSP tour cost.
- **Service times:** set $s_i=0$ for every customer and for the depot.
- **Time windows:** assign $[0,H]$ to every customer and to the depot, with

$$
H=mC_{\max}+1,\qquad C_{\max}=\max_{(i,j)\in E}c_{ij}.
$$

A Hamiltonian cycle has exactly $m$ arcs, each costing at most $C_{\max}$; therefore its total travel time is at most $mC_{\max}<H$. These wide windows do not remove any TSP tour.

- **Objective bound:** keep the same bound $B$.

The constructed instance contains a polynomial number of nodes and arcs. Because the integer $H$ has polynomially many bits relative to the input encoding, this transformation runs in polynomial time.

### 3.3. Proof of reduction

We prove equivalence in both directions.

**Direction 1: TSP-Decision has a solution $\Rightarrow$ CVRPTW-Decision has a solution.**

Suppose that the TSP instance has a Hamiltonian cycle of total cost at most $B$. Use this cycle as the route of the single vehicle. Every customer is visited exactly once; capacity $Q=m-1$ allows all unit demands; service durations are zero; and the time horizon is large enough to accommodate the complete route. Since travel times equal the original TSP costs, the total CVRPTW travel time is also at most $B$. Therefore, the constructed CVRPTW instance is a **yes-instance**.

**Direction 2: CVRPTW-Decision has a solution $\Rightarrow$ TSP-Decision has a solution.**

Suppose that the constructed CVRPTW instance admits a feasible solution of total travel time at most $B$. There is only one vehicle, and the routing rules require it to start from the depot, serve every customer exactly once and return to the depot. Consequently, its route is a Hamiltonian cycle through all vertices of the original TSP graph. Because the travel times were copied from TSP arc costs, the original cycle has cost at most $B$. Therefore, the TSP instance is also a **yes-instance**.

We have established:

$$
\boxed{\text{TSP-Decision is a yes-instance}\iff\text{CVRPTW-Decision is a yes-instance}.}
$$

Hence:

$$
\text{TSP-Decision}\leq_p\text{CVRPTW-Decision}.
$$

Since TSP-Decision is NP-complete, CVRPTW-Decision is NP-hard. Combined with membership in NP, this yields:

$$
\boxed{\text{CVRPTW-Decision is NP-complete}.}
$$

Consequently, the **optimization version of the general CVRPTW is NP-hard**.

## 4. Implications for finding a solution for the problem

The **NP-hardness** of the CVRPTW has several practical consequences:

- **No polynomial-time exact algorithm is expected in the general case unless $P=NP$.** A general-purpose algorithm cannot be assumed to return a proven optimum efficiently for every instance.
- **Exact methods may become computationally expensive.** Mixed-integer optimization can certify optimality on some instances, but the worst-case effort can increase rapidly as problem size grows.
- **Heuristics and metaheuristics are useful for larger instances.** They aim to find feasible, high-quality solutions within an acceptable computation time, without usually proving global optimality.
- **The method should depend on instance size and structure.** Small instances are useful for exact validation, while larger ones motivate methods such as ALNS.
- **Feasibility remains mandatory.** A short route that violates a capacity limit or a time window is not a valid CVRPTW solution.

These implications justify studying both a rigorous MILP formulation and a scalable heuristic. A computational comparison is still necessary to demonstrate the practical performance of the proposed method.

# Solution Method

##1. Choice of Method and Justification

We propose an Adaptive Large Neighborhood Search (ALNS) approach
for the Capacitated Vehicle Routing Problem with Time Windows
(CVRPTW).

The method combines three components:

1. A feasible insertion heuristic to construct an initial solution.
2. Destroy and repair operators to explore alternative customer
   assignments and visiting sequences.
3. A Simulated Annealing acceptance rule to allow occasional
   non-improving moves during the search.

The objective is to minimize total travel time while serving every
customer exactly once and respecting vehicle capacities, customer
time windows, the available fleet size and the depot return deadline.

## 2. Definition of a Solution

### 2.1 Route representation

Let $N=\{1,\ldots,n\}$ be the set of customers, let $0$ denote
the depot, and let $K$ be the set of available vehicles.

A complete solution is represented by a collection of non-empty
vehicle routes:

$$
S=(R_1,\ldots,R_m),
\qquad 1 \leq m \leq |K|.
$$

Each used vehicle performs one route:

$$
R_k=(v_{k,0},v_{k,1},\ldots,v_{k,r_k},v_{k,r_k+1}),
$$

where:

- $v_{k,0}=v_{k,r_k+1}=0$: the route starts and ends at the depot.
- $v_{k,p}\in N$ for $p=1,\ldots,r_k$: internal nodes are customers.
- $r_k$ is the number of customers served by route $k$.
- $m$ is the number of vehicles used.

Every customer must appear exactly once across all routes:

$$
\sum_{k=1}^{m}\sum_{p=1}^{r_k}
\mathbf{1}\{v_{k,p}=i\}=1,
\qquad \forall i\in N.
$$

Here, $\mathbf{1}\{\cdot\}$ equals 1 when the condition is true
and 0 otherwise.

For example, a solution serving five customers with two vehicles is:

$$
R_1=(0,2,5,0),
\qquad
R_2=(0,1,3,4,0).
$$

The route sequences determine both the assignment of customers
to vehicles and their visiting order. Service-start times are
computed from these sequences using the scheduling procedure
described in the next subsection.

### 2.2 Objective function

The objective is to minimize total travel time:

$$
\min_{S\in\mathcal{F}} Z(S)
=
\sum_{k=1}^{m} C(R_k),
$$

where $C(R_k)$ is the total travel time of route $k$,
including departure from and return to the depot.
$\mathcal{F}$ is the set of complete feasible solutions.

Waiting and service durations affect feasibility but
are not included in the objective.

### 2.3 Partial solutions

During the destroy step, some customers are temporarily
removed from the routes. Let $U$ be the set of customers
awaiting reinsertion.

The repair step attempts to reinsert every removed customer
while respecting all constraints.

A candidate solution is considered for acceptance only when
$U=\varnothing$ and all routes are feasible. Otherwise,
the candidate is rejected and the current solution is retained.


## 3. Route Feasibility

A route is feasible if it meets the vehicle's capacity, fits within customer time windows, and returns to the depot before the deadline.
It must also respect the time windows and the depot's return deadline.

### 3.1 Capacity constraint

The total demand on route $R_k$ cannot be more than
the vehicle's capacity $Q$:

$$
L(R_k)=\sum_{p=1}^{r_k}d_{v_{k,p}}\leq Q,
